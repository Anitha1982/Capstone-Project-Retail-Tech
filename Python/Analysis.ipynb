#### 3. Outliers

```python
numeric_columns = customers_df.select_dtypes(include=np.number).columns

outlier_report = []

for column in numeric_columns:

    Q1 = customers_df[column].quantile(0.25)
    Q3 = customers_df[column].quantile(0.75)

    IQR = Q3 - Q1

    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR

    outlier_count = (
        (customers_df[column] < lower_bound) |
        (customers_df[column] > upper_bound)
    ).sum()

    outlier_report.append({
        "Column": column,
        "Q1": Q1,
        "Q3": Q3,
        "IQR": IQR,
        "Lower_Bound": lower_bound,
        "Upper_Bound": upper_bound,
        "Outlier_Count": outlier_count
    })

outlier_report = pd.DataFrame(outlier_report)

print("\nOutlier Report:")
print(outlier_report)
```

## Document Data Attributes and Metadata

```python

def get_attribute_type(series):
    if pd.api.types.is_datetime64_any_dtype(series):
        return "Date/Time"
    elif pd.api.types.is_numeric_dtype(series):
        return "Numerical"
    elif pd.api.types.is_string_dtype(series):
        return "Categorical/Text"
    else:
        return "Other"


def create_metadata(df, dataset_name):

    return pd.DataFrame({
        "Dataset": dataset_name,
        "Column": df.columns,
        "Data_Type": df.dtypes.astype(str).values,
        "Attribute_Type": [
            get_attribute_type(df[col])
            for col in df.columns
        ],
        "Non_Null_Count": df.notna().sum().values,
        "Missing_Values": df.isna().sum().values,
        "Missing_Percentage": (
            df.isna().mean().values * 100
        ).round(2),
        "Unique_Values": df.nunique().values
    })


# Create metadata
metadata_df = pd.concat([
    create_metadata(customers_df, "Customers"),
    create_metadata(products_df, "Products"),
    create_metadata(orders_df, "Orders"),
    create_metadata(order_items_df, "Order Items"),
    create_metadata(order_reviews_df, "Order Reviews"),
    create_metadata(geo_df, "Geolocation")
], ignore_index=True)


# Save to Excel
metadata_df.to_excel(
    "Data_Metadata.xlsx",
    index=False
)

print("Data_Metadata.xlsx created successfully!")
```

## Address Outliers Using Statistical or Logical Methods

```python
order_items_df_clean = order_items_df.copy()

order_items_df_clean.loc[order_items_df_clean["unit_price"] <= 0, "unit_price"] = np.nan
order_items_df_clean["unit_price"] = order_items_df_clean["unit_price"].fillna(
    order_items_df_clean["unit_price"].median()
)

order_items_df_clean.loc[
    (order_items_df_clean["discount(%)"] < 0) |
    (order_items_df_clean["discount(%)"] > 100),
    "discount(%)"
] = np.nan
order_items_df_clean["discount(%)"] = order_items_df_clean["discount(%)"].fillna(
    order_items_df_clean["discount(%)"].median()
)

order_items_df_clean.loc[order_items_df_clean["shipping_cost"] < 0, "shipping_cost"] = np.nan
order_items_df_clean["shipping_cost"] = order_items_df_clean["shipping_cost"].fillna(
    order_items_df_clean["shipping_cost"].median()
)
```

#### Check the result

```python
print("Quantity invalid:",
      (order_items_df_clean["quantity"] <= 0).sum())

print("Unit price invalid:",
      (order_items_df_clean["unit_price"] <= 0).sum())

print("Discount invalid:",
      ((order_items_df_clean["discount(%)"] < 0) |
       (order_items_df_clean["discount(%)"] > 100)).sum())

print("Shipping cost invalid:",
      (order_items_df_clean["shipping_cost"] < 0).sum())
```

## Document Null and Alternate hypotheses clearly

# Null and Alternative Hypotheses

The hypotheses are formulated from the refined business questions and are designed to be tested using the statistical methods covered in the project: **T-test, Z-test, and Chi-square test**.

| Hypothesis                             | Null Hypothesis (H₀)                                                                                  | Alternative Hypothesis (H₁)                                                                  | Test       |
| -------------------------------------- | ----------------------------------------------------------------------------------------------------- | -------------------------------------------------------------------------------------------- | ---------- |
| **H1 – Customer Value**                | There is **no significant difference** in average spending between repeat and one-time customers.     | Repeat customers have **significantly higher average spending** than one-time customers.     | T-test     |
| **H2 – Purchase Frequency**            | There is **no significant difference** in average spending between frequent and infrequent customers. | Frequent customers have **significantly higher average spending** than infrequent customers. | T-test     |
| **H3 – Retention and Recency**         | There is **no significant difference** in average recency between active and churned customers.       | Active and churned customers have a **significantly different average recency**.             | T-test     |
| **H4 – Customer Type and Conversion**  | There is **no significant difference** in conversion rates between new and returning customers.       | Conversion rates are **significantly different** between new and returning customers.        | Z-test     |
| **H5 – Traffic Source and Conversion** | Customer conversion is **independent of traffic source**.                                             | Customer conversion is **significantly associated with traffic source**.                     | Chi-square |
| **H6 – Customer Segment and Churn**    | Customer churn is **independent of customer segment**.                                                | Customer churn is **significantly associated with customer segment**.                        | Chi-square |

## Decision Rule

All hypotheses will be evaluated at a **5% significance level (α = 0.05)**.

* **If p-value < 0.05:** Reject H₀ and support H₁.
* **If p-value ≥ 0.05:** Fail to reject H₀.

The final conclusion should be based on the statistical test results and should be translated into a **business interpretation**, rather than reporting only whether the hypothesis was accepted or rejected.

### Example

**Business Question:** Do repeat customers spend more than one-time customers?

* **H₀:** There is no significant difference in average spending between repeat and one-time customers.
* **H₁:** Repeat customers have significantly higher average spending than one-time customers.
* **Test:** Independent two-sample T-test
* **Significance Level:** α = 0.05
* **Decision:** Compare the calculated p-value with 0.05.
* **Business Interpretation:** Determine whether repeat customers should be prioritized through customer retention and loyalty strategies.


## Expected Direction and Tail Type



For each hypothesis, document whether the expected relationship is directional or non-directional, and select the appropriate tail of the statistical test.

| Hypothesis                             | Expected Direction                                               | Tail Type                           | Reason                                                                                                                                                  |
| -------------------------------------- | ---------------------------------------------------------------- | ----------------------------------- | ------------------------------------------------------------------------------------------------------------------------------------------------------- |
| **H1 – Repeat Customers & Spending**   | Repeat customers **spend more** than one-time customers.         | **One-tailed (right-tailed)**       | The assumption predicts a specific direction: higher spending.                                                                                          |
| **H2 – Purchase Frequency & Spending** | Frequent customers **spend more** than infrequent customers.     | **One-tailed (right-tailed)**       | The assumption predicts higher spending for frequent customers.                                                                                         |
| **H3 – Retention & Recency**           | Active and churned customers have **different** recency values.  | **Two-tailed**                      | No specific direction is assumed; only that the averages differ.                                                                                        |
| **H4 – Customer Type & Conversion**    | New and returning customers have **different** conversion rates. | **Two-tailed**                      | The hypothesis does not specify which group has the higher conversion rate.                                                                             |
| **H5 – Traffic Source & Conversion**   | Conversion is **associated with** traffic source.                | **Chi-square test of independence** | Chi-square tests whether the categorical variables are associated; it does not use one-/two-tailed direction in the same way as a mean/proportion test. |
| **H6 – Customer Segment & Churn**      | Churn is **associated with** customer segment.                   | **Chi-square test of independence** | The hypothesis tests whether the two categorical variables are related.                                                                                 |
